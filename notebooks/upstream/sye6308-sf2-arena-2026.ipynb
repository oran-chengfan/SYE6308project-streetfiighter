## 3. Run your own submission's actual self-check
Review the submitted Python and serialized weights first, then enable **both** `RUN_SELF_CHECK` and `TRUST_SUBMISSIONS`. A run switch requests execution; a trust switch acknowledges code loading. Neither switch means the model is safe or strong.

The arena loads **two independent copies of your selected model**. With the published offsets `[0, 12]`, it runs four fresh rounds: A/P1–B/P2 at offset 0, A/P1–B/P2 at offset 12, then A/P2–B/P1 at offsets 0 and 12. Here A and B are copies of the same agent. P1/P2 are controller ports, not permanent left/right screen positions.

An offset is a number of neutral native game frames before policy decisions begin. It is **not** another checkpoint, another training seed, or a round count. Each chosen action is a legal six-native-frame plan. Both actors see the same pre-decision game state; the emulator advances only after both replies arrive. Actors use CPU inference; the arena does not require a training GPU.

### Troubleshooting
| Symptom | What to check |
|---|---|
| Blank or wrong submission path | Attach the complete outer ZIP and copy its actual current path. No automatic model selection is performed. |
| Resource setup fails | A matching ROM is required; `.state` does not substitute. Read the exact Python/dependency/ROM error. |
| `ARENA_RULE_MISMATCH` or rules hash mismatch | Confirm the package and public arena release agree. Do not disable the check. |
| Static pass, then import/load error | Packaging checks do not verify runtime dependencies or model tensors. Inspect training versions and actor stderr. |
| `ARENA_CHECK_FAILED` / watchdog | Keep the diagnostic ZIP. This is a technical failure, not a scored loss in this draft. |
| Many `cutoff` results | The interface may work, but the rounds did not finish normally before the safety cap. Inspect the videos/inputs; do not report cutoffs as wins. |
| Model is weaker at P2 | Check its actual input and training coverage. P2 coordinates are self-first; the image is not automatically mirrored. |
