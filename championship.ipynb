# Counter-Ryu championship replay

Run this notebook from the repository root in the course environment with the game installed. It uses the released F and G weights. Each policy starts every bout with a fresh GRU state.

```python
from pathlib import Path
import sys

# If your notebook runs elsewhere, set REPO to the cloned repository.
REPO = Path.cwd()
if not (REPO / 'models/G/Group_10.zip').is_file():
    raise FileNotFoundError('Set REPO to the directory containing models/ and counter_ryu/.')
sys.path.insert(0, str(REPO))
from counter_ryu.play import play
```

## Qualification

Ryu faces Chun-Li at two stars and Ryu at two, four and eight stars. Each KO win earns 176 points plus the remaining health difference. Both released agents win all four by KO at the standard opening.

```python
for agent in ('F', 'G'):
    score = 0
    for character, stars in [('ChunLi', 2), ('Ryu', 2), ('Ryu', 4), ('Ryu', 8)]:
        bout = play(p1=agent, cpu=character, stars=stars)
        print(agent, character, stars, bout)
        if bout['winner'] != 'p1' or bout['reason'] != 'signed_hp_ko':
            raise RuntimeError('Unexpected qualification outcome. Inspect the game configuration.')
        score += 176 + bout['hp'][0] - bout['hp'][1]
    print(agent, 'qualification points', score)
```

## Swapped-port PvP

Run both assignments at the two Arena opening offsets. Equal opening delays apply to both players.

```python
for offset in (0, 12):
    for p1, p2 in [('G', 'F'), ('F', 'G')]:
        print('offset', offset, play(p1=p1, p2=p2, offset=offset))
```
